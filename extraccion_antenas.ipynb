```python
import pandas as pd
```

```python
# --- CONFIGURACIÓN ---

nombre_archivo_original = 'datos_consolidados_ordenados.csv'
nombre_archivo_nuevo = 'lista_antenas.csv'
```

```python
try:
    # Paso 1: Cargar el archivo CSV en un DataFrame de Pandas
    df = pd.read_csv(nombre_archivo_original)

    # Paso 2: Extraer la columna y limpiar el texto.
    # Usamos .str.replace() para quitar la parte específica del texto.
    antenas_limpias = df['fuente_archivo'].str.replace('-air-quality.csv', '', regex=False)

    # Paso 3: Obtener los valores únicos para no tener nombres repetidos.
    antenas_unicas = antenas_limpias.unique()

    # Paso 4: Crear un nuevo DataFrame con los nombres únicos y el encabezado deseado.
    df_resultado = pd.DataFrame(antenas_unicas, columns=['antena'])

    # Paso 5: Guardar el nuevo DataFrame en un archivo CSV.
    # `index=False` evita que Pandas añada una columna extra con números de fila.
    df_resultado.to_csv(nombre_archivo_nuevo, index=False)

    print(f" Excelente Se ha creado el archivo '{nombre_archivo_nuevo}' con {len(df_resultado)} antenas únicas.")

except FileNotFoundError:
    print(f" Error: No se encontró el archivo '{nombre_archivo_original}'. Asegúrate de que el nombre sea correcto y esté en la misma carpeta.")
except KeyError:
    print(" Error: No se encontró la columna 'fuente_archivo' en el CSV. Revisa que el nombre de la columna sea exacto.")
```
